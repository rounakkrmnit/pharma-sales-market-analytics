```python
import pandas as pd
import numpy as np

df = pd.read_csv("pharma-data[1].csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
print("Shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
print("Negative Quantity:", (df["Quantity"] < 0).sum())
print("Negative Price:", (df["Price"] < 0).sum())
print("Negative Sales:", (df["Sales"] < 0).sum())

print("\nNumerical summary:")
display(df[["Quantity", "Price", "Sales"]].describe())
```

```python
duplicates = df[df.duplicated(keep=False)]

print("Duplicate records:")
display(duplicates.sort_values(list(df.columns)))
```

```python
negative_sales = df[df["Quantity"] < 0]

print("Negative transaction rows:", len(negative_sales))

display(
    negative_sales[
        ["Customer Name", "Country", "Channel", "Product Name",
         "Product Class", "Quantity", "Price", "Sales",
         "Month", "Year", "Sales Team"]
    ].head(20)
)
```

```python
print("Negative transactions by year:")
display(negative_sales.groupby("Year").size())

print("\nNegative transactions by channel:")
display(negative_sales["Channel"].value_counts())

print("\nNegative transactions by product class:")
display(negative_sales["Product Class"].value_counts())
```

```python
df = df.drop_duplicates().copy()

print("Shape after removing duplicates:", df.shape)
print("Remaining duplicates:", df.duplicated().sum())
```

```python
negative_sales = df[df["Quantity"] < 0]

print("Negative transactions:", len(negative_sales))
print("Negative sales value:", negative_sales["Sales"].sum())
print("Positive sales value:", df[df["Sales"] > 0]["Sales"].sum())

print(
    "Negative sales as % of positive sales:",
    round(
        abs(negative_sales["Sales"].sum()) /
        df[df["Sales"] > 0]["Sales"].sum() * 100,
        2
    ),
    "%"
)
```

```python
negative_by_year = (
    negative_sales.groupby("Year")
    .agg(
        Transactions=("Sales", "size"),
        Return_Sales=("Sales", "sum")
    )
    .reset_index()
)

negative_by_year
```

```python
# Remove exact duplicate records
clean_df = df.drop_duplicates().copy()

print("Original rows:", len(df))
print("Cleaned rows:", len(clean_df))
print("Duplicates remaining:", clean_df.duplicated().sum())
```

```python
# Verify Sales = Quantity × Price
clean_df["Calculated_Sales"] = clean_df["Quantity"] * clean_df["Price"]

print(
    "Sales calculation matches:",
    np.isclose(
        clean_df["Sales"],
        clean_df["Calculated_Sales"]
    ).all()
)

clean_df.drop(columns="Calculated_Sales", inplace=True)
```

```python
annual_sales = clean_df.groupby("Year").agg(
    Net_Sales=("Sales", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Transactions=("Sales", "count")
).reset_index()

annual_sales["YoY_Growth_%"] = (
    annual_sales["Net_Sales"].pct_change() * 100
).round(2)

annual_sales
```

```python
product_class = clean_df.groupby("Product Class").agg(
    Net_Sales=("Sales", "sum"),
    Quantity=("Quantity", "sum"),
    Transactions=("Sales", "count")
).sort_values("Net_Sales", ascending=False).reset_index()

product_class["Sales_Contribution_%"] = (
    product_class["Net_Sales"] / clean_df["Sales"].sum() * 100
).round(2)

product_class
```

```python
channel_analysis = clean_df.groupby("Channel").agg(
    Net_Sales=("Sales", "sum"),
    Quantity=("Quantity", "sum"),
    Transactions=("Sales", "count")
).sort_values("Net_Sales", ascending=False).reset_index()

channel_analysis["Sales_Contribution_%"] = (
    channel_analysis["Net_Sales"] / clean_df["Sales"].sum() * 100
).round(2)

channel_analysis
```

```python
product_analysis = clean_df.groupby("Product Name").agg(
    Net_Sales=("Sales", "sum"),
    Quantity=("Quantity", "sum"),
    Transactions=("Sales", "count")
).sort_values("Net_Sales", ascending=False).reset_index()

product_analysis["Sales_Contribution_%"] = (
    product_analysis["Net_Sales"] / clean_df["Sales"].sum() * 100
).round(2)

product_analysis.head(10)
```

```python
country_analysis = clean_df.groupby("Country").agg(
    Net_Sales=("Sales", "sum"),
    Quantity=("Quantity", "sum"),
    Transactions=("Sales", "count")
).sort_values("Net_Sales", ascending=False).reset_index()

country_analysis["Sales_Contribution_%"] = (
    country_analysis["Net_Sales"] / clean_df["Sales"].sum() * 100
).round(2)

country_analysis.head(10)
```

```python
return_analysis = clean_df.groupby("Year").agg(
    Total_Sales=("Sales", "sum"),
    Return_Value=("Sales", lambda x: x[x < 0].sum()),
    Return_Transactions=("Sales", lambda x: (x < 0).sum()),
    Total_Transactions=("Sales", "count")
).reset_index()

return_analysis["Return_Rate_%"] = (
    return_analysis["Return_Transactions"] /
    return_analysis["Total_Transactions"] * 100
).round(2)

return_analysis["Return_Value_%_of_Sales"] = (
    abs(return_analysis["Return_Value"]) /
    return_analysis["Total_Sales"].abs() * 100
).round(2)

return_analysis
```

```python
# Monthly Sales Analysis

month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

monthly_sales = clean_df.groupby("Month").agg(
    Net_Sales=("Sales", "sum"),
    Transactions=("Sales", "count")
).reset_index()

monthly_sales["Month"] = pd.Categorical(
    monthly_sales["Month"],
    categories=month_order,
    ordered=True
)

monthly_sales = monthly_sales.sort_values("Month")

monthly_sales["Sales_Contribution_%"] = (
    monthly_sales["Net_Sales"] / clean_df["Sales"].sum() * 100
).round(2)

monthly_sales
```

```python
customer_analysis = clean_df.groupby("Customer Name").agg(
    Net_Sales=("Sales", "sum"),
    Transactions=("Sales", "count")
).sort_values("Net_Sales", ascending=False).reset_index()

customer_analysis["Sales_Contribution_%"] = (
    customer_analysis["Net_Sales"] / clean_df["Sales"].sum() * 100
).round(2)

customer_analysis.head(10)
```

```python
team_analysis = clean_df.groupby("Sales Team").agg(
    Net_Sales=("Sales", "sum"),
    Transactions=("Sales", "count"),
    Quantity=("Quantity", "sum")
).sort_values("Net_Sales", ascending=False).reset_index()

team_analysis["Sales_Contribution_%"] = (
    team_analysis["Net_Sales"] / clean_df["Sales"].sum() * 100
).round(2)

team_analysis
```

```python
rep_analysis = clean_df.groupby("Name of Sales Rep").agg(
    Net_Sales=("Sales", "sum"),
    Transactions=("Sales", "count"),
    Quantity=("Quantity", "sum")
).sort_values("Net_Sales", ascending=False).reset_index()

rep_analysis["Sales_Contribution_%"] = (
    rep_analysis["Net_Sales"] / clean_df["Sales"].sum() * 100
).round(2)

rep_analysis.head(10)
```

```python
kpi_summary = {
    "Total_Net_Sales": clean_df["Sales"].sum(),
    "Total_Transactions": len(clean_df),
    "Total_Quantity": clean_df["Quantity"].sum(),
    "Product_Classes": clean_df["Product Class"].nunique(),
    "Products": clean_df["Product Name"].nunique(),
    "Countries": clean_df["Country"].nunique(),
    "Customers": clean_df["Customer Name"].nunique(),
    "Sales_Reps": clean_df["Name of Sales Rep"].nunique(),
    "Sales_Teams": clean_df["Sales Team"].nunique()
}

kpi_summary
```

```python
import matplotlib.pyplot as plt

annual_sales = (
    clean_df.groupby("Year")["Sales"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(8, 5))
plt.plot(annual_sales["Year"], annual_sales["Sales"], marker="o")
plt.title("Annual Net Sales")
plt.xlabel("Year")
plt.ylabel("Net Sales")
plt.grid(True)
plt.show()
```

```python
product_class_sales = (
    clean_df.groupby("Product Class")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))
product_class_sales.plot(kind="bar")
plt.title("Sales by Product Class")
plt.xlabel("Product Class")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
channel_sales = (
    clean_df.groupby("Channel")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(7, 5))
channel_sales.plot(kind="bar")
plt.title("Sales by Channel")
plt.xlabel("Channel")
plt.ylabel("Net Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
